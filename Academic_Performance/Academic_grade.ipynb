```python
fig, axes = plt.subplots(5, 2, figsize=(14, 16))
axes = axes.flatten()
numeric_cols = ['age', 'gaming_hours', 'study_hours', 'sleep_hours',
       'attendance', 'social_activity', 'device_usage',
       'reaction_time_ms', 'addiction_score']
for idx, col in enumerate(numeric_cols):
    sns.histplot(df[col], kde=True, ax=axes[idx])
    axes[idx].set_title(f'Distribution of {col}')

# Hide the last empty subplot
axes[-1].set_visible(False)

plt.tight_layout()
plt.show()
```

### Key insights from the distributions
- Age: narrow adult range → low variance; likely limited predictive power alone.
- Gaming hours: heavy right-skew / many low values → consider log/boxcox or binning; potential nonlinear relationship with grades.
- Study hours: central tendency with spread → likely an important predictor.
- Sleep hours / Attendance: may show ceiling effects (clustered at high values) — limited discriminative power unless small differences matter.
- Social activity / Device usage / Addiction score: often skewed and may contain outliers; likely informative but may need transformation.
- Reaction time: wide range and outliers → normalize or winsorize before modeling.
- Overall: several numeric features show skewness and/or outliers; some features may have low variance.

### Suggested actions for feature selection
- Preprocessing
    - Handle outliers (winsorize or trim) and apply transforms for skewed features (log, sqrt, box-cox).
    - Scale numeric features (StandardScaler / RobustScaler).
    - Encode categorical (one-hot or target encoding for high-cardinality).
- Univariate screening
    - Compute Pearson/Spearman correlations with grades and ANOVA for categorical features; drop extremely low-correlation features.
- Multicollinearity
    - Calculate VIF; remove or combine highly collinear features.
- Model-based selection
    - Use tree-based feature importances (RandomForest / XGBoost) or L1-regularized linear models (Lasso) to rank features.
    - Use permutation importance to assess true impact on model performance.
- Wrapper methods
    - Recursive Feature Elimination (RFE) with cross-validation to find optimal subset.
- Dimensionality reduction (if many correlated features)
    - PCA for orthogonal components (useful for modeling but reduces interpretability).
- Validation
    - Always validate selected features via cross-validated model performance (e.g., CV RMSE) to avoid overfitting.
- Domain knowledge
    - Prioritize features with known academic relevance (study_hours, attendance) even if statistical signals are modest.

Recommended pipeline: preprocess (outliers, transform, encode, scale) → quick filter (correlation/VIF) → model-based ranking (RF/XGBoost or Lasso) → RFE/permutation importance with CV → finalize features based on performance + domain knowledge.

```python
from sklearn.model_selection import train_test_split
```

```python

X_train, X_test, y_train, y_test = train_test_split(x_inp, y, test_size=0.2, random_state=42)

X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

```python
categorical_cols = [
    'gender',
    'gaming_genre',
    'stress_level'
]

numeric_cols = [
    'age',
    'gaming_hours',
    'study_hours',
    'sleep_hours',
    'attendance',
    'social_activity',
    'device_usage',
    'reaction_time_ms',
    'addiction_score'
]
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
```

```python

from sklearn.preprocessing import (
    OneHotEncoder,
    StandardScaler
)

from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error
```

```python
# -----------------------------------
# Preprocessing
# -----------------------------------
preprocessor = ColumnTransformer(
    transformers=[

        # Scale numeric columns
        ('num', StandardScaler(), numeric_cols),

        # Encode categorical columns
        ('cat', OneHotEncoder(drop='first'), categorical_cols)
    ]
)
```

```python
# -----------------------------------
# Full pipeline
# -----------------------------------
model = Pipeline([
    ('preprocessing', preprocessor),
    ('regressor', LinearRegression())
])
```

```python
# -----------------------------------
# Train model
# -----------------------------------
model.fit(X_train, y_train)
```

```python
X_train.describe()
```

```python
# -----------------------------------
# Prediction
# -----------------------------------
y_pred = model.predict(X_test)
(y_pred - y_test).mean()
```

```python

# -----------------------------------
# Evaluation
# -----------------------------------
print("R2 Score:", r2_score(y_test, y_pred))
print("MSE:", mean_squared_error(y_test, y_pred))
```

```python
import joblib

joblib.dump(model, 'student_grade_predictor.pkl')
```
