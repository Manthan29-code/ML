```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python

# 1. Load the dataset
# Assuming the file is in the same directory as your script
df = pd.read_csv('customers.csv')

# Set the visual style for seaborn
sns.set_theme(style="whitegrid")

fig, axes = plt.subplots(2, 2, figsize=(12, 10))
fig.suptitle('Raw Data Visualization Before Cleaning', fontsize=16, fontweight='bold')

# --- Plot 1: Scatter Plot (Top-Left) ---
sns.scatterplot(
    data=df, 
    x='AverageSpend', 
    y='AverageFrequency', 
    s=100, 
    alpha=0.7,
    ax=axes[0, 0]
)
axes[0, 0].set_title('Scatter Plot: Spend vs. Frequency')
axes[0, 0].set_xlabel('Average Spend')
axes[0, 0].set_ylabel('Average Frequency')

# --- Plot 2: Histogram for AverageSpend (Top-Right) ---
sns.histplot(
    df['AverageSpend'], 
    kde=True, 
    bins=15, 
    color='skyblue', 
    ax=axes[0, 1]
)
axes[0, 1].set_title('Distribution of Average Spend')
axes[0, 1].set_xlabel('Average Spend')

# --- Plot 3: Histogram for AverageFrequency (Bottom-Left) ---
sns.histplot(
    df['AverageFrequency'], 
    kde=True, 
    bins=15, 
    color='salmon', 
    ax=axes[1, 0]
)
axes[1, 0].set_title('Distribution of Average Frequency')
axes[1, 0].set_xlabel('Average Frequency')

# --- Plot 4: Boxplots (Bottom-Right) ---
sns.boxplot(
    data=df[['AverageSpend', 'AverageFrequency']], 
    orient='h', 
    palette='Set2',
    ax=axes[1, 1]
)
axes[1, 1].set_title('Boxplots (Checking Data Spread & Outliers)')

# Adjust layout so plots do not overlap
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```
