```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python

# -----------------------------
# Load Dataset
# -----------------------------
df = pd.read_csv("Iris.csv")

# -----------------------------
# Drop ID Column
# -----------------------------
df = df.drop(columns=["Id"])

# Set plotting style
sns.set_theme(style="whitegrid")
```

```python

# ======================================================
# 1. Dataset Information
# ======================================================

print(df.head())
print(df.info())
```

```python

# ======================================================
# 2. Class Distribution
# ======================================================

plt.figure(figsize=(6,4))
sns.countplot(data=df, x="Species")
plt.title("Species Distribution")
plt.show()
```

```python

# ======================================================
# 3. Histograms
# Understand distribution of every feature
# ======================================================

df.hist(
    figsize=(10,8),
    bins=20,
    edgecolor="black",
)

plt.suptitle("Feature Distributions")
plt.show()
```

```python

# ======================================================
# 4. Boxplots
# Detect Outliers
# ======================================================

plt.figure(figsize=(12,6))

for i, column in enumerate(df.columns[:-1], 1):
    plt.subplot(2,2,i)
    sns.boxplot(y=df[column])
    plt.title(column)

plt.tight_layout()
plt.show()
```

```python

# ======================================================
# 5. Correlation Heatmap
# ======================================================

plt.figure(figsize=(8,6))

sns.heatmap(
    df.drop(columns=["Species"]).corr(),
    annot=True,
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Matrix")
plt.show()
```

```python

# ======================================================
# 6. Pair Plot
# Shows relationship between every feature
# ======================================================

sns.pairplot(
    df,
    hue="Species",
    diag_kind="hist"
)

plt.show()
```

```python

# ======================================================
# 7. Scatter Plot
# Petal Length vs Petal Width
# ======================================================

plt.figure(figsize=(7,5))

sns.scatterplot(
    data=df,
    x="PetalLengthCm",
    y="PetalWidthCm",
    hue="Species",
    s=70
)

plt.title("Petal Length vs Petal Width")
plt.show()
```

```python

# ======================================================
# 8. Scatter Plot
# Sepal Length vs Sepal Width
# ======================================================

plt.figure(figsize=(7,5))

sns.scatterplot(
    data=df,
    x="SepalLengthCm",
    y="SepalWidthCm",
    hue="Species",
    s=70
)

plt.title("Sepal Length vs Sepal Width")
plt.show()
```

```python

# ======================================================
# 9. Violin Plots
# Compare feature distributions by species
# ======================================================

plt.figure(figsize=(12,8))

for i, column in enumerate(df.columns[:-1], 1):
    plt.subplot(2,2,i)

    sns.violinplot(
        data=df,
        x="Species",
        y=column
    )

    plt.xticks(rotation=15)
    plt.title(column)

plt.tight_layout()
plt.show()
```

```python

# ======================================================
# 10. KDE Distribution
# ======================================================

plt.figure(figsize=(8,6))

for species in df["Species"].unique():

    subset = df[df["Species"] == species]

    sns.kdeplot(
        subset["PetalLengthCm"],
        label=species,
        fill=True,
        alpha=0.3
    )

plt.title("Petal Length Distribution")
plt.legend()
plt.show()
```
